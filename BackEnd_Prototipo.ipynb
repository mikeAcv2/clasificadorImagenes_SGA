```python
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np
import os
import PIL
import tensorflow as tf
from keras.preprocessing import image
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential
from google.colab import drive
import matplotlib.image as mpimg

drive.mount('/content/drive')

#etiquetado de imagenes
type_label=['corrosion', 'environment', 'exclamasion_mark', 'exploding_bomb', 'flame', 'flame_over_circle', 'gas_cylinder', 'health_hazard', 'skull_and_crossbones']

#etiquetado de bahias
container_label = ['BA01', 'BA02', 'BA03']


# se debe crear una carpeta en "Mi Unidad" con nombre "Simulado",despues crear un acceso directo de la carpeta "Prototipo" y
# la carpeta "trabajoFinal", que se encuentran en "Compartido conmigo", ambos accesos directos
# deben ser depositados en la carpeta creada "Simulado"
direccion_raiz = '/content/drive/MyDrive/Simulado/trabajoFinal'
validation_dir = direccion_raiz + '/entrega2'
direccion_originales  = direccion_raiz + '/imagenesOriginales/'
direccion_modelo = direccion_raiz + '/modelo/'

# Parámetros
batch_size = 1
img_height = 150
img_width = 150


# carga del modelo
new_model = tf.keras.models.load_model(direccion_modelo+'best_model_weights.h5');
new_model.compile(
  optimizer='adam',
  loss=tf.losses.SparseCategoricalCrossentropy(from_logits=True),
  metrics=['accuracy'])

# Create two textarea widgets
textarea1 = widgets.Textarea(
    value='',
    placeholder='Ingre el numero de folio...',
    description='No. de folio:',
    layout=widgets.Layout(width='50%', height='100px')
)


# Create a button widget
button = widgets.Button(
    description='Validar',
    button_style='success'
)

# Define the button click event
def on_button_click(b):
    print("Textarea 1 content:")
    print(textarea1.value)
    paraNivelBusquda = textarea1.value
    if  paraNivelBusquda:
       paraNivelBusquda = paraNivelBusquda + '/'

    directorioFolios = direccion_raiz + '/Folios2/' + paraNivelBusquda
    new_row=[]

    for root, dirs, files in os.walk(directorioFolios):
        path = root.split(os.sep)
        new_row = [0] * len(type_label)
        for file in files:
            #print(new_row)
            if (file.endswith(".png")):
    #            file_anadir = directorioFolios + "\\" + os.path.basename(root) + "\\" + file
                file_anadir = directorioFolios  + file
                #print(file_anadir)
                if os.path.exists(file_anadir) and file.endswith(".png"):
                    # predicting images
                    img = image.load_img(file_anadir, target_size=(img_width, img_height))
                    x = image.img_to_array(img)
                    x = np.expand_dims(x, axis=0)
                    images = np.vstack([x])
                    classes = new_model.predict(images, batch_size=batch_size)
                    #print(classes)
                    for idxr,row in enumerate(classes):
                        compara = -9999
                        for idx,value in enumerate(row):
                          if compara < value:
                            compara = value
                            idx_max =  idx
                        new_row[idx_max] = new_row[idx_max]+1

    zero_count = sum(1 for item in new_row if item > 0)
    print(new_row)
    print(f"Number of null elements: {zero_count}")
    for idx, x in enumerate(new_row):
      if x != 0:
        print(type_label[idx])
        direccion_imagen = direccion_originales + type_label[idx] + '.png'
        img = mpimg.imread(direccion_imagen)
        plt.figure()
        #contenedor vs etiqueta
        if idx < 3:
           subPlot_label = 0
        elif idx >= 3 and idx < 6:
           subPlot_label = 1
        else:
           subPlot_label = 2
        plt.title(container_label[subPlot_label])
        plt.axis('off')
        plt.imshow(img)



button.on_click(on_button_click)

# Display the widgets
display(textarea1, button)

```
